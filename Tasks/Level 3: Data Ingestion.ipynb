```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType

schema = StructType([
    StructField("order_id", IntegerType(), True),
    StructField("customer_name", StringType(), True),
    StructField("amount", StringType(), True),
    StructField("status", StringType(), True)
])

data = [
    (1, "Ali", "1200.50", "completed"),
    (2, "Sara", "800.00", "pending"),
    (3, "John", "invalid_amount", "completed")
]

df = spark.createDataFrame(data, schema=schema)

output_path = "/Volumes/lab/dev/lab/users.csv"
(
    df.write
    .option("header", "true")
    .option("delimiter", ";")
    .mode("overwrite")
    .csv(output_path)
)
```

```python
df = (
    spark.read
    .option("header", "true")
    .option("delimiter", ";")
    .option("inferSchema", "true")
    .csv(output_path)
)

display(df)
```
