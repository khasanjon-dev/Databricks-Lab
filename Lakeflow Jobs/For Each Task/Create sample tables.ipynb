```python
customers = [
    (1, "Ali", "Tashkent"),
    (2, "Sara", "Samarkand"),
    (3, "John", "Bukhara")
]

orders = [
    (101, 1, 250.0),
    (102, 2, 150.0),
    (103, 1, 400.0)
]

products = [
    (1, "Laptop", 1200.0),
    (2, "Phone", 800.0),
    (3, "Tablet", 400.0)
]

customers_df = spark.createDataFrame(
    customers, ["customer_id", "name", "city"]
)

orders_df = spark.createDataFrame(
    orders, ["order_id", "customer_id", "amount"]
)

products_df = spark.createDataFrame(
    products, ["product_id", "product_name", "price"]
)

customers_df.write.mode("overwrite").saveAsTable("lab.dev.customers")
orders_df.write.mode("overwrite").saveAsTable("lab.dev.orders")
products_df.write.mode("overwrite").saveAsTable("lab.dev.products")
```
