```python
from pyspark.sql import functions as F

df = spark.table("lab.dev.orders_bronze")
valid_condition = (
    F.col("order_id").isNotNull() &
    F.col("customer_id").isNotNull() &
    (F.col("amount") > 0)
    )

invalid_df = df.filter(~valid_condition)

invalid_count = invalid_df.count()

if invalid_count > 0:
    print(f"Validation Error: Found {invalid_count} invalid record(s):")
    invalid_df.show()
    raise ValueError(f"Job Task Failed: {invalid_count} invalid records detected in lab.dev.orders.")
else:
    print("Data validation passed successfully! No invalid records found.")
```
