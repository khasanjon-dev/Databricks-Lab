```python
from pyspark.sql import functions as F

df = spark.table("lab.dev.orders_bronze")


valid_condition = (
    F.col("order_id").isNotNull()
    & F.col("customer_id").isNotNull()
    & F.col("amount").isNotNull()
    & (F.col("amount") > 0)
)

valid_df = df.filter(valid_condition)
invalid_df = df.filter(~valid_condition)

valid_df.write \
    .format("delta") \
        .mode("overwrite") \
            .saveAsTable("lab.dev.orders_silver")

invalid_df.write \
    .format("delta") \
        .mode("overwrite") \
            .saveAsTable("lab.dev.orders_quarantine")

print(f"Valid records: {valid_df.count()}")
print(f"Invalid records: {invalid_df.count()}")
```
