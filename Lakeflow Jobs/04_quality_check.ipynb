```python
from pyspark.sql import functions as F

df = spark.table("lab.dev.orders_gold")

invalid_condition = (
    F.col("order_id").isNull()
    | F.col("customer_id").isNull()
    | F.col("amount").isNull()
    | (F.col("amount") <= 0)
    | F.col("processed_at").isNull()
)

if df.isEmpty():
    raise ValueError("Quality check failed: Gold table is empty.")

invalid_count = df.filter(invalid_condition).count()

if invalid_count > 0:
    raise ValueError(
        f"Quality check failed: {invalid_count} invalid records "
        "in lab.dev.orders_gold."
    )

print("Quality check passed.")
```
