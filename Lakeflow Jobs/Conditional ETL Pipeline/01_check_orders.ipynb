```python
from pyspark.sql import functions as F

df = spark.table("lab.dev.orders_bronze")

condition = (
    F.col("order_id").isNotNull()
    & F.col("customer_id").isNotNull()
    & (F.col("amount") > 0)
)

validated_df = df.filter(condition)

valid_count = validated_df.count()

dbutils.jobs.taskValues.set("valid_order_count", valid_count)

validated_df.write \
    .format("delta") \
        .mode("overwrite") \
            .saveAsTable("lab.dev.silver")
```
