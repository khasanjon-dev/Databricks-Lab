```python
from pyspark.sql import functions as F

df = spark.table("lab.dev.orders_silver")

gold_df = df.withColumn(
    "processed_at",
    F.current_timestamp()
)

if gold_df.isEmpty():
    raise ValueError("Transformation failed: Silver table contains no records.")

gold_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("lab.dev.orders_gold")

print("Transformation completed. Gold table updated.")
```
