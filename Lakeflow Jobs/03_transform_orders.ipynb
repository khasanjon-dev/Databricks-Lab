```python
from pyspark.sql import functions as F


df = spark.table("lab.dev.orders_silver")

condition = (
    F.col("order_id").isNotNull()
    & F.col("customer_id").isNotNull()
    & (F.col("amount") > 0)
)

validated_df = df.filter(condition)


validated_df = validated_df.withColumn("processed_at", F.current_timestamp())

count_df = validated_df.count()

if count_df >= 1:
    validated_df.write \
        .format("delta") \
            .mode("overwrite") \
                .saveAsTable("orders_silver")

else:
    print("No data!")
    raise ValueError("No data")
```
