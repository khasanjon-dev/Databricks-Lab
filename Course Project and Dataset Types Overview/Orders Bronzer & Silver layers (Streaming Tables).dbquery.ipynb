```sql
CREATE OR REFRESH STREAMING TABLE dev.bronze.orders
AS
SELECT
    *
FROM STREAM read_files(
    "/Volumes/dev/landing/orders_json",
    format => "json"
);

SELECT * FROM dev.bronze.orders;


CREATE OR REFRESH STREAMING TABLE dev.silver.orders
AS
SELECT
    order_id, 
    CAST(amount AS DECIMAL(10, 2)) AS amount,
    status,
    customer_id,
    CAST(timestamp AS TIMESTAMP) AS order_timestamp,
    source_file,
    processing_time
FROM STREAM(dev.bronze.orders)
WHERE order_id IS NOT NULL
    AND customer_id IS NOT NULL
    AND status IS NOT NULL;


SELECT * FROM dev.silver.orders;
```
