```sql
CREATE OR REFRESH MATERIALIZED VIEW dev.gold.full_order_info
AS
SELECT
    orders.order_id,
    orders.customer_id,
    orders.amount,
    orders.order_timestamp,
    status.status,
    status.status_timestamp
FROM dev.silver.orders AS orders
INNER JOIN dev.silver.orders_status as status
ON orders.order_id = status.order_id
;

SELECT * FROM dev.gold.full_order_info;


CREATE OR REFRESH MATERIALIZED VIEW dev.gold.cancelled_orders
AS
SELECT
    *
FROM dev.gold.full_order_info
WHERE status = 'cancelled';


SELECT * FROM dev.gold.cancelled_orders;


CREATE OR REFRESH MATERIALIZED VIEW dev.gold.delivered_orders
AS
SELECT
    *
FROM dev.gold.full_order_info
WHERE status = 'delivered';


SELECT * FROM dev.gold.delivered_orders;



SELECT COUNT(*) FROM dev.gold.full_order_info;

SELECT COUNT(*) FROM dev.gold.cancelled_orders;

SELECT COUNT(*) FROM dev.gold.delivered_orders;
```
