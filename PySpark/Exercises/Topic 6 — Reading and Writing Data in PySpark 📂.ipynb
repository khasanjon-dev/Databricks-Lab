```python
orders_data = [
    (1, 101, 250.0, "completed"),
    (2, 102, 150.0, "pending"),
    (3, 101, 300.0, "completed"),
    (4, 103, 100.0, "cancelled"),
    (5, 102, 400.0, "completed")
]

columns = ["order_id", "customer_id", "amount", "status"]

orders_df = spark.createDataFrame(orders_data, columns)
```

```python
# 1
(
    orders_df.write
    .mode("overwrite")
    .parquet("/Volumes/dev/landing/test_files/orders")
)
# 2
orders_parquet_df = (
    spark.read.parquet("/Volumes/dev/landing/parquet_files/orders")
)
orders_parquet_df.display()
# 3
(
    orders_df.write
    .format("csv")
    .mode("overwrite")
    .option("header", True)
    .csv("/Volumes/dev/landing/test_files/orders.csv")
)
orders_csv_df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv("/Volumes/dev/landing/test_files/orders.csv")
)
orders_csv_df.display()
# 4
(
    orders_df.write
    .partitionBy("status")
    .parquet("/Volumes/dev/landing/test_files/orders_partioned")
)

# 5
(
    orders_df.write
    .format("delta")
    .saveAsTable("dev.landing.orders_learning")
)


# 6
orders_delta_df = (
    spark.read.table("dev.landing.orders_learning")
)

orders_delta_df.display()
```
