```python
customers_data = [
    (101, "Ali", "Uzbekistan"),
    (102, "John", "USA"),
    (103, "Sarah", "UK"),
    (104, "David", "Germany"),
    (105, "Emma", "Canada"),
    (106, "Michael", "France")
]

orders_data = [
    (1, 101, 250.0, "completed"),
    (2, 102, 150.0, "pending"),
    (3, 101, 300.0, "completed"),
    (4, 107, 450.0, "cancelled"),
    (5, 103, 200.0, "completed"),
    (6, 104, 500.0, "completed"),
    (7, 107, 100.0, "pending")
]

customers_df = spark.createDataFrame(
    customers_data,
    ["customer_id", "customer_name", "country"]
)

orders_df = spark.createDataFrame(
    orders_data,
    ["order_id", "customer_id", "amount", "status"]
)
```

```python
# Part A — Joins

result_1 = customers_df.join(
    orders_df,
    on="customer_id",
    how="inner"
)

result_2 = customers_df.join(
    orders_df,
    on="customer_id",
    how="left"
)

result_3 = customers_df.join(
    orders_df,
    on="customer_id",
    how="full_outer"
)

result_4 = orders_df.join(
    customers_df,
    on="customer_id",
    how="left_anti"
)

result_5 = customers_df.join(
    orders_df,
    on="customer_id",
    how="left_semi"
)
```

```python
# Part B — Aggregation after Join
from pyspark.sql.functions import col, sum

result_6 = (
    customers_df.alias("c")
    .join(orders_df.alias("o"), on="customer_id", how="inner")
    .groupBy("c.customer_id")
    .agg(sum(col("o.amount")).alias("total_amount"))
    .filter(col("total_amount") > 400)
)
result_6.display()
```

```python
# Part C — Union
orders_other_data = [
    (1, 101, "completed", 250.0),
    (2, 102, "pending", 150.0),
    (3, 101, "completed", 300.0),
    (4, 107, "cancelled", 450.0),
    (5, 103, "completed", 200.0),
    (6, 104, "completed", 500.0),
    (7, 107, "pending", 100.0),
]


orders_other_df = spark.createDataFrame(
    orders_other_data, ["order_id", "customer_id", "status", "amount"]
)

result_9 = orders_other_df.unionByName(
    orders_df
)

result_9.display()
```
