### 1. Prepare our datasets

```python
customers_data = [
    (101, "Ali", "Uzbekistan"),
    (102, "John", "USA"),
    (103, "Sarah", "UK"),
    (104, "David", "Germany"),
    (105, "Emma", "Canada")
]

customers_df = spark.createDataFrame(
    customers_data,
    ["customer_id", "customer_name", "country"]
)
```

```python
orders_data = [
    (1, 101, 250.0),
    (2, 102, 150.0),
    (3, 101, 300.0),
    (4, 106, 450.0),
    (5, 103, 200.0)
]

orders_df = spark.createDataFrame(
    orders_data,
    ["order_id", "customer_id", "amount"]
)
```

### 2. Inner Join

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="inner"
)

result.display()
```

### 3. Left Join

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="left"
)

result.display()
```

### 4. Right Join

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="right"
)

result.display()
```

### 5. Full Outer Join

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="full_outer"
)

result.display()
```

### 6. Left Semi Join

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="left_semi"
)

result.display()
```

### 7. Left Anti Join

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="left_anti"
)

result.display()
```

### 9. Union vs. UnionByName

```python
orders_january = [
    (1, 101, 250.0),
    (2, 102, 150.0)
]

orders_february = [
    (3, 103, 300.0),
    (4, 104, 450.0)
]

df_january = spark.createDataFrame(
    orders_january,
    ["order_id", "customer_id", "amount"]
)

df_february = spark.createDataFrame(
    orders_february,
    ["order_id", "customer_id", "amount"]
)
```

```python
all_orders = df_january.union(df_february)

all_orders.display()
```

```python
all_orders = df_january.unionByName(df_february)

all_orders.display()
```

### allowMissingColumns

```python
df1 = spark.createDataFrame(
    [(1, "Ali")],
    ["id", "name"]
)

df2 = spark.createDataFrame(
    [(2, "John", "IT")],
    ["id", "name", "department"]
)
```

```python
result = df1.unionByName(
    df2,
    allowMissingColumns=True
)

result.show()
```

### 10. Handling Duplicate Column Names

```python
from pyspark.sql import functions as F

orders_alias = orders_df.alias("o")
customers_alias = customers_df.alias("c")

result = orders_alias.join(
    customers_alias,
    F.col("o.customer_id") == F.col("c.customer_id"),
    "inner"
)

result.display()
```

### Solution 1: Use the column name directly

```python
result = orders_df.join(
    customers_df,
    on="customer_id",
    how="inner"
)

result.display()
```

### Solution 2: Explicitly select columns

```python
result = (
    orders_alias
    .join(
        customers_alias,
        F.col("o.customer_id") == F.col("c.customer_id"),
        "inner"
    )
    .select(
        F.col("o.order_id"),
        F.col("o.customer_id"),
        F.col("o.amount"),
        F.col("c.customer_name"),
        F.col("c.country")
    )
)

result.display()
```

### 11. Quick Reference
Here are all the join types you should remember.
Join type	What it returns
- inner	Matching records from both
- left	All left + matching right
- right	All right + matching left
- full_outer	All records from both
- left_semi	Matching records from left only
- left_anti	Non-matching records from left only
- cross	Every possible combination
