### 1. Reading data

```python
df = (
    spark.read
    .format("json")
    .option("inferSchema", True)
    .load("/Volumes/dev/landing/orders_json")
)

df.display()
```

### 2. Defining an explicit schema

```python
from pyspark.sql.types import (
    StructType,
    StructField,
    IntegerType,
    DoubleType,
    StringType
)

schema = StringType([
    StructField("order_id", IntegerType(), False),
    StructField("customer_id", IntegerType(), True),
    StructField("amount", DoubleType(), True),
    StructField("status", StringType(), True)
])

df = (
    spark.read
    .format("csv")
    .option("header", True)
    .schema(schema)
    .load("/Volumes/dev/landing/orders_json")
)
```
