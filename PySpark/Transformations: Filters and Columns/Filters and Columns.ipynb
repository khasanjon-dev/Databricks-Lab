```python
data = [
    (1, "Laptop", 1200.0, "completed"),
    (2, "Phone", 800.0, "pending"),
    (3, "Tablet", 400.0, "completed"),
    (4, "Monitor", 300.0, "pending"),
    (5, "Keyboard", 100.0, "cancelled")
]

schema = ["order_id", "product", "amount", "status"]

df = spark.createDataFrame(data, schema=schema)

display(df)
```

Example 1: Add a constant column

```python
from pyspark.sql.functions import lit

df_new = df.withColumn("country", lit("Uzbekistan"))

display(df_new)
```

Example 2: Calculate a column from existing columns

```python
df_new = df.withColumn(
    "tax",
    df.amount * 0.10
)

df_new = df_new.withColumn(
    "total_amount",
    df_new.amount + df_new.tax
)

display(df_new)
```

Example 3: Add a conditional column

```python
from pyspark.sql.functions import when

df_new = df.withColumn(
    "order_category",
    when(df.amount >= 1000, "premium")
    .when(df.amount >= 500, "standart")
    .otherwise("budget")
)

display(df_new)
```

4. Replacing an existing column

```python
from pyspark.sql.functions import upper

df_updated = df.withColumn(
    "status",
    upper(df.status)
)

display(df_updated)
```

Another example: Replace a column with a conditional expression

```python
from pyspark.sql.functions import when

df_updated = df.withColumn(
    "status",
    when(df.amount < 300, "review")
    .otherwise(df.status)
)

display(df_updated)
```

5. Renaming and dropping columns

```python
df_renamed = df.withColumnRenamed(
    "amount",
    "order_amount"
)

df_renamed.printSchema()
```

Remove a column

```python
df_dropped = df.drop("status", "product")

display(df_dropped)
```

6. Combining filters and column transformations

```python
from pyspark.sql.functions import col, lit

result_df = (
    df
    .filter(col("status") == "completed")
    .withColumn("tax", col("amount") * lit(0.10))
    .withColumn("total_amount", col("amount") + col("tax"))
    .filter(col("total_amount") > 500)
)

display(df)
display(result_df)
```
