```python
data = [
    (1, "Laptop", 1200.0, "completed"),
    (2, "Phone", 800.0, "pending"),
    (3, "Tablet", 400.0, "completed"),
    (4, "Monitor", 300.0, "pending"),
    (5, "Keyboard", 100.0, "cancelled")
]

schema = ["order_id", "product", "amount", "status"]

df = spark.createDataFrame(data, schema=schema)

display(df)
```

Task 1 — Basic filtering

```python
df_filtered = df.filter(df.amount > 300)

display(df_filtered)
```

Task 2 — Multiple conditions

```python
from pyspark.sql.functions import col

df.filter(
    (col("amount") >= 300) & (col("status") == "pending")
).show()
```

Task 3 — Filter with isin()

```python
df.filter(col("status").isin("completed", "cancelled")).show()
```

Task 4 — Add a constant column

```python
from pyspark.sql.functions import lit

result = df.withColumn("currency", lit("USD"))

result.show()
```

Task 5 — Calculate a new column

```python
result = df.withColumn("discount", col("amount") * 0.05)

result.show()
```

Task 6 — Add a conditional column

```python
from pyspark.sql.functions import when

result = df.withColumn(
    "priority",
    when(col("amount") >= 1000, "high")
    .when(col("amount") >= 400, "medium")
    .otherwise("low")
)

result.show()
```

Task 7 — Replace an existing column

```python
from pyspark.sql.functions import upper, col

result = df.withColumn("status", upper(col("status")))

result.show()
```

Task 8 — Rename and drop columns
1.

```python
df.show()

result = df.withColumnRenamed("amount", "order_amount")
result = result.drop("status")

result.show()
```

Task 9 — Combine everything

```python
new_df = df.filter(col("status") == "completed")

new_df = new_df.withColumn("tax", col("amount") * 0.1)

new_df = new_df.withColumn("total_amount", col("amount") + col("tax"))

display(new_df)
```
