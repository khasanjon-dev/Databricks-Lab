```python
import re

df = (
    spark.read
    # .option("multiLine", "true")
    .json("/Volumes/lab/dev/lab/users.json")
)

def clean_column_name(name):
    name = re.sub(r"[ ,;{}()\n\t=]+", "_", name)
    return name.strip("_")

for column in df.columns:
    df = df.withColumnRenamed(column, clean_column_name(column))

display(df)
```
