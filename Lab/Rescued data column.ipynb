```python
from pyspark.sql.types import (
    StructType,
    StructField,
    LongType,
    StringType
)

file_path = "/Volumes/lab/dev/sales_files/users_mismatched.csv"

schema = StructType([
    StructField("user_id", LongType(), True),
    StructField("name", StringType(), True),
    StructField("age", LongType(), True)
])

df = (
    spark.read
    .format("csv")
    .option("header", "true")
    .option("rescuedDataColumn", "_rescued_data")
    .schema(schema)
    .load(file_path)
)

display(df)
```
