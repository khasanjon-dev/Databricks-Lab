```sql
CREATE OR REPLACE TABLE main.ingestion_lab.raw_orders AS
SELECT * FROM VALUES
    (101, 'completed', 120.50),
    (102, 'pending',    75.00),
    (103, 'completed', 240.00),
    (104, 'cancelled',  30.00)
AS t(order_id, status, amount);

SELECT * FROM main.ingestion_lab.raw_orders;

CREATE OR REPLACE TABLE main.ingestion_lab.completed_orders AS
SELECT * FROM main.ingestion_lab.raw_orders
WHERE status == 'completed';

SELECT 
    COUNT(order_id),
    SUM(amount)
FROM main.ingestion_lab.completed_orders;
```
