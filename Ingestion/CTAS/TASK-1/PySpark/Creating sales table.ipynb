creating

```python
data = [
    (1, 'Ali', 120.00, 'completed', 'Uzbekistan'),
    (2, 'John', 250.00, 'pending', 'USA'),
    (3, 'Sara', 180.00, 'completed', 'UK'),
    (4, 'Akmal', 90.00, 'cancelled', 'Uzbekistan'),
    (5, 'Maria', 300.00, 'completed', 'USA')
]

columns = ["order_id", "customer_name", "amount", "status", "country"]

df = spark.createDataFrame(data, columns)

df.write.format("delta").mode("overwrite").saveAsTable("lab.dev.orders_raw")
```

checking

```python
df = spark.table("lab.dev.orders_raw")
df.show()
```
