SPARK and SQL

```python
volume_path = "/Volumes/main/ingestion_lab/raw_files/"
source_path = volume_path + "orders.csv"
```

```python
files = dbutils.fs.ls(volume_path)
display(files)
```

```python
for file in files:
    print(file.name)
```

```python
df = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(source_path)
)

display(df.limit(5))
```

```python
df.printSchema()
```

Saving data to table

```python
target_table = "main.ingestion_lab.orders_from_spark"

(
    df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable(target_table)
)
```

Checking data quality

```python
result_df = spark.table(target_table)

display(result_df.limit(5))
```

```python
df = spark.sql(f"""
               SELECT *
               FROM {target_table}
               """)

display(df.limit(5))
```
